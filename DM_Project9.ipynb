```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
data = {
    "Product_ID": [1, 2, 3, 4, 5, 6, 7, 8],
    "Product": [
        "Laptop",
        "Gaming Laptop",
        "Smartphone",
        "Headphones",
        "Wireless Mouse",
        "Keyboard",
        "Smart Watch",
        "Bluetooth Speaker"
    ],
    "Category": [
        "Electronics",
        "Electronics",
        "Electronics",
        "Audio",
        "Computer Accessories",
        "Computer Accessories",
        "Wearable",
        "Audio"
    ],
    "Description": [
        "laptop computer office work",
        "gaming laptop computer high performance",
        "smartphone mobile phone android",
        "wireless headphones music audio",
        "wireless mouse computer accessory",
        "mechanical keyboard computer gaming",
        "smart watch fitness wearable",
        "bluetooth speaker wireless music"
    ]
}

df = pd.DataFrame(data)
```

```python
print("Product Dataset:")
print(df)
```

```python
df["Description"] = df["Description"].str.lower()
```

```python
df = df.dropna()
```

```python
print("\nAfter Preprocessing:")
print(df)
```

```python
print("\nProducts by Category:")
print(df["Category"].value_counts())
```

```python
vectorizer = TfidfVectorizer()
```

```python
tfidf_matrix = vectorizer.fit_transform(df["Description"])
```

```python
similarity = cosine_similarity(tfidf_matrix)
```

```python
def recommend_product(product_name, number=3):

    if product_name not in df["Product"].values:
        print("Product not found.")
        return
    index = df[df["Product"] == product_name].index[0]

    scores = list(enumerate(similarity[index]))

    scores = sorted(
        scores,
        key=lambda x: x[1],
        reverse=True
    )

    print("\nRecommendations for:", product_name)
    print("--------------------------------")

    count = 0

    for i, score in scores:


        if i == index:
            continue

        print(
            df.iloc[i]["Product"],
            "-",
            df.iloc[i]["Category"],
            "- Similarity:",
            round(score, 2)
        )

        count += 1

        if count == number:
            break
```

```python
recommend_product("Laptop", 3)

recommend_product("Headphones", 3)

recommend_product("Smartphone", 3)
```
